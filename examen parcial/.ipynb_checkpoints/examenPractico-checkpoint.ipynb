## Examen Practico - Segundo Parcial

#### 0) Cargar una de las imágenes histológicas

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from skimage import io, color, filters, morphology, measure
from scipy import ndimage
import cv2
from PIL import Image
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

# Cargar la imagen en formato RGB
img = io.imread('histo_1.jpg')

# Normalizar la imagen al rango [0, 1]
if img.dtype == np.uint8:
    img_normalized = img.astype(np.float32) / 255.0
else:
    img_normalized = img.astype(np.float32)

print(f"Dimensiones de la imagen: {img_normalized.shape}")
print(f"Rango de valores: [{img_normalized.min():.3f}, {img_normalized.max():.3f}]")

# Visualizar la imagen
plt.figure(figsize=(10, 8))
plt.imshow(img_normalized)
plt.title('Imagen Histológica Original (RGB)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

#### 1) Realizar una transformación de color para convertir la imagen al espacio de color CMYK

```python
# Convertir de RGB a CMYK
def rgb_to_cmyk(img_rgb):
    """Convierte una imagen RGB a CMYK"""
    img_rgb = np.clip(img_rgb, 0, 1)
    K = 1 - np.max(img_rgb, axis=2)
    C = (1 - img_rgb[:, :, 0] - K) / (1 - K + 1e-8)
    M = (1 - img_rgb[:, :, 1] - K) / (1 - K + 1e-8)
    Y = (1 - img_rgb[:, :, 2] - K) / (1 - K + 1e-8)
    return np.dstack([C, M, Y, K])

img_cmyk = rgb_to_cmyk(img_normalized)
img_magenta = img_cmyk[:, :, 1]  # Extraer componente magenta

# Visualizar la imagen del canal magenta
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_normalized)
plt.title('Imagen Original (RGB)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_magenta, cmap='gray')
plt.title('Canal Magenta (CMYK)')
plt.axis('off')
plt.tight_layout()
plt.show()

print(f"Rango del canal magenta: [{img_magenta.min():.3f}, {img_magenta.max():.3f}]")
```

#### 2) Umbralizar la imagen para separar los píxeles del fondo de la región tisular

```python
# Aplicar filtro gaussiano
img_magenta_uint8 = (img_magenta * 255).astype(np.uint8)
img_blurred = cv2.GaussianBlur(img_magenta_uint8, (5, 5), 0)

# Aplicar umbralización de Otsu
threshold, binary_mask = cv2.threshold(img_blurred, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# Convertir a binario (0 y 1)
binary_mask = binary_mask.astype(np.uint8)

print(f"Umbral de Otsu utilizado: {threshold}")
print(f"Pixeles blancos (lumen + fondo): {np.sum(binary_mask > 0)}")

# Visualizar la máscara resultante
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_blurred, cmap='gray')
plt.title('Imagen Filtrada (Gaussiano 5x5)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_mask, cmap='gray')
plt.title('Máscara Binaria (Otsu)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

#### 3) Limpiar la imagen eliminando los artefactos de lumen (objetos blancos pequeños que no son lúmenes)

```python
# Eliminar objetos pequeños (artefactos de lumen menores a 300 píxeles)
cleaned_mask = morphology.remove_small_objects(binary_mask.astype(bool), min_size=300)
cleaned_mask = cleaned_mask.astype(np.uint8)

print(f"Píxeles después de limpiar artefactos: {np.sum(cleaned_mask > 0)}")
print(f"Píxeles eliminados: {np.sum(binary_mask) - np.sum(cleaned_mask)}")

# Visualizar la máscara resultante
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(binary_mask, cmap='gray')
plt.title('Máscara Binaria (con artefactos)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cleaned_mask, cmap='gray')
plt.title('Máscara Limpia (sin artefactos < 300px)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

#### 4) Rellenar con 0s el fondo de la imagen para quedarnos únicamente con los lúmenes

```python
# Función para encontrar automáticamente las semillas
def find_seeds_automatic(mask):
    """Encuentra automáticamente dos semillas: una en lumen y otra en fondo"""
    h, w = mask.shape
    
    # Semilla 1: en el interior de un lumen (píxel blanco cercano al centro)
    seed_lumen = None
    for y in range(h // 4, 3 * h // 4):
        for x in range(w // 4, 3 * w // 4):
            if mask[y, x] > 0:
                seed_lumen = (y, x)
                break
        if seed_lumen:
            break
    
    # Si no encontramos semilla en el centro, usar cualquier píxel blanco
    if seed_lumen is None:
        coords = np.argwhere(mask > 0)
        if len(coords) > 0:
            seed_lumen = tuple(coords[0])
    
    # Semilla 2: en el fondo (píxel negro en las esquinas)
    seed_background = None
    for y in [0, h-1]:
        for x in [0, w-1]:
            if mask[y, x] == 0:
                seed_background = (y, x)
                break
        if seed_background:
            break
    
    if seed_background is None:
        # Si no hay píxel negro en esquinas, buscar en los bordes
        for y in range(h):
            if mask[y, 0] == 0:
                seed_background = (y, 0)
                break
    
    return seed_lumen, seed_background

# Encontrar semillas automáticamente
seed_lumen, seed_background = find_seeds_automatic(cleaned_mask)
print(f"Semilla lumen: {seed_lumen}")
print(f"Semilla fondo: {seed_background}")

# Aplicar region growing usando flood fill
rg_mask = np.zeros_like(cleaned_mask)

# Flood fill desde la semilla de lumen (en blanco)
if seed_lumen:
    cv2.floodFill(cleaned_mask, None, seed_lumen, 255)
    rg_mask = cleaned_mask.copy()

# Invertir para que lumen sea 1 y fondo sea 0
rg_mask = (rg_mask > 0).astype(np.uint8)

print(f"Píxeles de lumen después de region growing: {np.sum(rg_mask)}")

# Visualizar la máscara resultante
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cleaned_mask, cmap='gray')
plt.title('Máscara Antes de Region Growing')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(rg_mask, cmap='gray')
plt.title('Máscara Después de Region Growing')
plt.axis('off')
plt.tight_layout()
plt.show()
```

#### 5) Rellenar los objetos de los lúmenes

```python
# Rellenar los lúmenes usando binary_fill_holes
filled_mask = ndimage.binary_fill_holes(rg_mask.astype(bool))
filled_mask = filled_mask.astype(np.uint8)

print(f"Píxeles después de rellenar: {np.sum(filled_mask)}")

# Visualizar la máscara resultante
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(rg_mask, cmap='gray')
plt.title('Máscara Antes de Rellenar')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(filled_mask, cmap='gray')
plt.title('Máscara Después de Rellenar (binary_fill_holes)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

#### 6) Detectar y dibujar los contornos de los lúmenes sobre la imagen original

```python
# Convertir la imagen normalizada a uint8 para dibujar contornos
img_uint8 = (img_normalized * 255).astype(np.uint8)

# Encontrar contornos en la máscara rellena
contours, hierarchy = cv2.findContours(filled_mask, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

# Dibujar los contornos en verde sobre la imagen original
img_with_contours = img_uint8.copy()
cv2.drawContours(img_with_contours, contours, -1, (0, 255, 0), 2)  # Verde: (0, 255, 0) en BGR

print(f"Número de lúmenes detectados: {len(contours)}")

# Visualizar la imagen superpuesta
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img_uint8, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_with_contours, cv2.COLOR_BGR2RGB))
plt.title('Contornos de Lúmenes (Verde)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

#### 7) Identificar y cropear el lumen más grande

```python
# Encontrar el lumen de mayor área
max_area = 0
max_contour = None
max_index = -1

for i, contour in enumerate(contours):
    area = cv2.contourArea(contour)
    if area > max_area:
        max_area = area
        max_contour = contour
        max_index = i

print(f"Área del lumen más grande: {max_area:.2f} píxeles")

# Obtener el bounding box del lumen más grande
if max_contour is not None:
    x, y, w, h = cv2.boundingRect(max_contour)
    
    # Cropear el lumen sobre la imagen original
    lumen_crop = img_uint8[y:y+h, x:x+w].copy()
    
    print(f"Dimensiones del crop: {lumen_crop.shape}")
    print(f"Coordenadas del lumen (x, y, w, h): ({x}, {y}, {w}, {h})")
    
    # Visualizar el lumen cropeado
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img_with_contours, cv2.COLOR_BGR2RGB))
    rect = plt.Rectangle((x, y), w, h, fill=False, edgecolor='red', linewidth=2)
    plt.gca().add_patch(rect)
    plt.title('Imagen con Lumen Más Grande Seleccionado')
    plt.axis('off')
    
    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(lumen_crop, cv2.COLOR_BGR2RGB))
    plt.title('Lumen Cropeado (Imagen Original)')
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

#### 8) Extraer 13 características geométricas que permitan caracterizar el lumen recortado 

```python
# Extraer la máscara binaria del lumen más grande
lumen_mask = np.zeros((h, w), dtype=np.uint8)
cv2.drawContours(lumen_mask, [max_contour], 0, 1, -1)

# Obtener las propiedades del lumen usando regionprops
labeled = measure.label(lumen_mask)
props = measure.regionprops(labeled)[0]

# Calcular las 13 características geométricas
features = {}

# 1) Área
features['Area'] = round(props.area, 4)

# 2) Área de la bounding box
features['Area_bbox'] = round(props.bbox_area, 4)

# 3) Área convexa
features['Area_convexa'] = round(props.convex_area, 4)

# 4) Excentricidad
features['Excentricidad'] = round(props.eccentricity, 4)

# 5) Diámetro equivalente
features['Diametro_equivalente'] = round(props.equivalent_diameter, 4)

# 6) Extensión (Extent)
features['Extension'] = round(props.extent, 4)

# 7) Diámetro Feret (máxima distancia entre dos puntos del contorno)
feret_diameter = 0
for i in range(len(max_contour)):
    for j in range(i+1, len(max_contour)):
        dist = np.linalg.norm(max_contour[i] - max_contour[j])
        if dist > feret_diameter:
            feret_diameter = dist
features['Diametro_Feret'] = round(float(feret_diameter), 4)

# 8) Longitud del eje mayor
features['Longitud_eje_mayor'] = round(props.major_axis_length, 4)

# 9) Longitud del eje menor
features['Longitud_eje_menor'] = round(props.minor_axis_length, 4)

# 10) Orientación (en radianes)
features['Orientacion'] = round(props.orientation, 4)

# 11) Perímetro
contour_length = cv2.arcLength(max_contour, True)
features['Perimetro'] = round(float(contour_length), 4)

# 12) Solidez (Solidity = Area / Convex Area)
features['Solidez'] = round(props.solidity, 4)

# 13) Compacidad (4π * Area / Perímetro²)
if contour_length > 0:
    compactness = (4 * np.pi * props.area) / (contour_length ** 2)
else:
    compactness = 0
features['Compacidad'] = round(float(compactness), 4)

# Mostrar todas las características
print("=" * 60)
print("CARACTERÍSTICAS GEOMÉTRICAS DEL LUMEN MÁS GRANDE")
print("=" * 60)
for i, (name, value) in enumerate(features.items(), 1):
    print(f"{i:2d}. {name:25s}: {value:12.4f}")
print("=" * 60)

# Guardar en un DataFrame para mejor visualización
import pandas as pd
df_features = pd.DataFrame(list(features.items()), columns=['Característica', 'Valor'])
print("\nTabla de características:")
print(df_features.to_string(index=False))

# Crear una visualización de las características
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Subplot 1: Imagen del lumen
axes[0, 0].imshow(cv2.cvtColor(lumen_crop, cv2.COLOR_BGR2RGB))
axes[0, 0].set_title('Lumen Cropeado')
axes[0, 0].axis('off')

# Subplot 2: Máscara binaria del lumen
axes[0, 1].imshow(lumen_mask[y:y+h, x:x+w], cmap='gray')
axes[0, 1].set_title('Máscara Binaria del Lumen')
axes[0, 1].axis('off')

# Subplot 3: Elipse de eje mayor y menor
ellipse = patches.Ellipse(
    xy=props.centroid,
    width=2*props.minor_axis_length/2,
    height=2*props.major_axis_length/2,
    angle=np.degrees(props.orientation),
    fill=False,
    edgecolor='red',
    linewidth=2
)
axes[1, 0].imshow(img_uint8[y:y+h, x:x+w])
axes[1, 0].set_title('Elipse de Eje Mayor y Menor')
axes[1, 0].axis('off')

# Subplot 4: Tabla de características pequeña
axes[1, 1].axis('off')
table_data = []
for name, value in list(features.items())[:6]:
    table_data.append([name, f"{value:.4f}"])
table = axes[1, 1].table(cellText=table_data, colLabels=['Característica', 'Valor'],
                         cellLoc='left', loc='center', bbox=[0, 0.5, 1, 1])
table.auto_set_font_size(False)
table.set_fontsize(9)
axes[1, 1].text(0.5, 0.3, f"Total características: 13", ha='center', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

# Guardar características en un diccionario para uso futuro
print("\nDiccionario de características:")
print(features)
```
