# PROCESAMIENTO DE IMAGEN 1 (histo_1.jpg)

### 0) Cargar la primera imagen histológica

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from skimage import io, color, filters, morphology, measure
from scipy import ndimage
import cv2
from PIL import Image
import pandas as pd
from math import pi
import warnings
warnings.filterwarnings('ignore')

# Cargar la imagen en formato RGB
img = io.imread('histo_1.jpg')

# Normalizar la imagen al rango [0, 1]
if img.dtype == np.uint8:
    img_normalized = img.astype(np.float32) / 255.0
else:
    img_normalized = img.astype(np.float32)

print(f"Dimensiones de la imagen: {img_normalized.shape}")
print(f"Rango de valores: [{img_normalized.min():.3f}, {img_normalized.max():.3f}]")

# Visualizar la imagen
plt.figure(figsize=(10, 8))
plt.imshow(img_normalized)
plt.title('Imagen Histológica 1 Original (RGB)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### 1) Transformación de color a CMYK

```python
# Convertir de RGB a CMYK
def rgb_to_cmyk(img_rgb):
    """Convierte una imagen RGB a CMYK"""
    img_rgb = np.clip(img_rgb, 0, 1)
    K = 1 - np.max(img_rgb, axis=2)
    C = (1 - img_rgb[:, :, 0] - K) / (1 - K + 1e-8)
    M = (1 - img_rgb[:, :, 1] - K) / (1 - K + 1e-8)
    Y = (1 - img_rgb[:, :, 2] - K) / (1 - K + 1e-8)
    return np.dstack([C, M, Y, K])

img_cmyk = rgb_to_cmyk(img_normalized)
img_magenta = img_cmyk[:, :, 1]  # Extraer componente magenta

# Visualizar la imagen del canal magenta
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_normalized)
plt.title('Imagen Original (RGB)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_magenta, cmap='gray')
plt.title('Canal Magenta (CMYK)')
plt.axis('off')
plt.tight_layout()
plt.show()

print(f"Rango del canal magenta: [{img_magenta.min():.3f}, {img_magenta.max():.3f}]")
```

### 2) Umbralización

```python
# Aplicar filtro gaussiano
img_magenta_uint8 = (img_magenta * 255).astype(np.uint8)
img_blurred = cv2.GaussianBlur(img_magenta_uint8, (5, 5), 0)

# Aplicar umbralización de Otsu
threshold, binary_mask = cv2.threshold(img_blurred, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

# Convertir a binario (0 y 1)
binary_mask = binary_mask.astype(np.uint8)

print(f"Umbral de Otsu utilizado: {threshold}")
print(f"Pixeles blancos (lumen + fondo): {np.sum(binary_mask > 0)}")

# Visualizar la máscara resultante
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_blurred, cmap='gray')
plt.title('Imagen Filtrada (Gaussiano 5x5)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_mask, cmap='gray')
plt.title('Máscara Binaria (Otsu)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### 3) Limpieza de artefactos

```python
# Eliminar objetos pequeños (artefactos de lumen menores a 300 píxeles)
cleaned_mask = morphology.remove_small_objects(binary_mask.astype(bool), min_size=300)
cleaned_mask = cleaned_mask.astype(np.uint8)

print(f"Píxeles después de limpiar artefactos: {np.sum(cleaned_mask > 0)}")
print(f"Píxeles eliminados: {np.sum(binary_mask) - np.sum(cleaned_mask)}")

# Visualizar la máscara resultante
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(binary_mask, cmap='gray')
plt.title('Máscara Binaria (con artefactos)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cleaned_mask, cmap='gray')
plt.title('Máscara Limpia (sin artefactos < 300px)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### 4) Region Growing

```python
# ─── REEMPLAZA TODO TU BLOQUE DE REGION GROWING CON ESTO ───

# Enfoque con componentes conectados (más robusto que flood fill para lúmenes)
h, w = cleaned_mask.shape

num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(
    cleaned_mask, connectivity=8
)

# Filtrar componentes por área (eliminar ruido y el fondo negro)
min_area = 100                # ignora ruido pequeño
max_area = (h * w) * 0.3     # ignora región negra gigante (tejido)

lumen_mask = np.zeros_like(cleaned_mask)
lumenes_encontrados = 0

for i in range(1, num_labels):  # i=0 es el fondo negro
    area = stats[i, cv2.CC_STAT_AREA]
    if min_area < area < max_area:
        lumen_mask[labels == i] = 255
        lumenes_encontrados += 1

print(f"Lúmenes encontrados: {lumenes_encontrados}")
print(f"Píxeles totales de lumen: {np.sum(lumen_mask > 0)}")

# Visualización
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cleaned_mask, cmap='gray')
plt.title('Máscara Original (cleaned_mask)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(lumen_mask, cmap='gray')
plt.title(f'Máscara de Lúmenes ({lumenes_encontrados} encontrados)')
plt.axis('off')

plt.tight_layout()
plt.show()

# Guardar como rg_mask para compatibilidad con el resto de tu código
rg_mask = lumen_mask.copy()
```

### 5) Relleno de lúmenes

```python
from scipy import ndimage
import cv2

# Rellenar huecos por componente individual (más controlado)
filled_mask = np.zeros_like(rg_mask)

num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(
    rg_mask, connectivity=8
)

for i in range(1, num_labels):
    componente = (labels == i).astype(np.uint8)
    componente_filled = ndimage.binary_fill_holes(componente).astype(np.uint8)
    filled_mask = cv2.bitwise_or(filled_mask, componente_filled * 255)

print(f"Lúmenes procesados: {num_labels - 1}")
print(f"Píxeles después de rellenar: {np.sum(filled_mask > 0)}")

# Visualizar
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(rg_mask, cmap='gray')
plt.title('Máscara Antes de Rellenar')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(filled_mask, cmap='gray')
plt.title('Máscara Después de Rellenar (por componente)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### 6) Detección de contornos

```python
# Convertir la imagen normalizada a uint8 para dibujar contornos
img_uint8 = (img_normalized * 255).astype(np.uint8)

# Encontrar contornos en la máscara rellena
contours, hierarchy = cv2.findContours(filled_mask, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

# Dibujar los contornos en verde sobre la imagen original
img_with_contours = img_uint8.copy()
cv2.drawContours(img_with_contours, contours, -1, (0, 255, 0), 2)  # Verde: (0, 255, 0) en BGR

print(f"Número de lúmenes detectados: {len(contours)}")

# Visualizar la imagen superpuesta
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img_uint8, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_with_contours, cv2.COLOR_BGR2RGB))
plt.title('Contornos de Lúmenes (Verde)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

### 7) Extracción del lumen más grande

```python
# Encontrar el lumen de mayor área
max_area = 0
max_contour = None
max_index = -1

for i, contour in enumerate(contours):
    area = cv2.contourArea(contour)
    if area > max_area:
        max_area = area
        max_contour = contour
        max_index = i

print(f"Área del lumen más grande: {max_area:.2f} píxeles")

# Obtener el bounding box del lumen más grande
if max_contour is not None:
    x, y, w, h = cv2.boundingRect(max_contour)
    
    # Cropear el lumen sobre la imagen original
    lumen_crop = img_uint8[y:y+h, x:x+w].copy()
    
    print(f"Dimensiones del crop: {lumen_crop.shape}")
    print(f"Coordenadas del lumen (x, y, w, h): ({x}, {y}, {w}, {h})")
    
    # Visualizar el lumen cropeado
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img_with_contours, cv2.COLOR_BGR2RGB))
    rect = plt.Rectangle((x, y), w, h, fill=False, edgecolor='red', linewidth=2)
    plt.gca().add_patch(rect)
    plt.title('Imagen con Lumen Más Grande Seleccionado')
    plt.axis('off')
    
    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(lumen_crop, cv2.COLOR_BGR2RGB))
    plt.title('Lumen Cropeado (Imagen Original)')
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```

### 8) Extracción de 13 características geométricas

```python
# Extraer la máscara binaria del lumen más grande
lumen_mask = np.zeros((h, w), dtype=np.uint8)
cv2.drawContours(lumen_mask, [max_contour], 0, 1, -1)

# Obtener las propiedades del lumen usando regionprops
labeled = measure.label(lumen_mask)
props = measure.regionprops(labeled)[0]

# Calcular las 13 características geométricas
features = {}

# 1) Área
features['Area'] = round(props.area, 4)

# 2) Área de la bounding box
features['Area_bbox'] = round(props.bbox_area, 4)

# 3) Área convexa
features['Area_convexa'] = round(props.convex_area, 4)

# 4) Excentricidad
features['Excentricidad'] = round(props.eccentricity, 4)

# 5) Diámetro equivalente
features['Diametro_equivalente'] = round(props.equivalent_diameter, 4)

# 6) Extensión (Extent)
features['Extension'] = round(props.extent, 4)

# 7) Diámetro Feret
feret_diameter = 0
for i in range(len(max_contour)):
    for j in range(i+1, len(max_contour)):
        dist = np.linalg.norm(max_contour[i] - max_contour[j])
        if dist > feret_diameter:
            feret_diameter = dist
features['Diametro_Feret'] = round(float(feret_diameter), 4)

# 8) Longitud del eje mayor
features['Longitud_eje_mayor'] = round(props.major_axis_length, 4)

# 9) Longitud del eje menor
features['Longitud_eje_menor'] = round(props.minor_axis_length, 4)

# 10) Orientación (en radianes)
features['Orientacion'] = round(props.orientation, 4)

# 11) Perímetro
contour_length = cv2.arcLength(max_contour, True)
features['Perimetro'] = round(float(contour_length), 4)

# 12) Solidez
features['Solidez'] = round(props.solidity, 4)

# 13) Compacidad
if contour_length > 0:
    compactness = (4 * np.pi * props.area) / (contour_length ** 2)
else:
    compactness = 0
features['Compacidad'] = round(float(compactness), 4)

# Mostrar todas las características
print("=" * 60)
print("CARACTERÍSTICAS GEOMÉTRICAS DEL LUMEN MÁS GRANDE - IMAGEN 1")
print("=" * 60)
for i, (name, value) in enumerate(features.items(), 1):
    print(f"{i:2d}. {name:25s}: {value:12.4f}")
print("=" * 60)

# Guardar en un DataFrame
df_features = pd.DataFrame(list(features.items()), columns=['Característica', 'Valor'])
print("\nTabla de características:")
print(df_features.to_string(index=False))
```

# ANÁLISIS COMPARATIVO ENTRE AMBAS IMÁGENES

```python
# Crear una tabla comparativa de características
comparison_data = []
for key in features.keys():
    comparison_data.append({
        'Característica': key,
        'Imagen 1': features[key],
        'Imagen 2': features_2[key],
        'Diferencia': round(features[key] - features_2[key], 4),
        'Diferencia %': round((abs(features[key] - features_2[key]) / ((features[key] + features_2[key])/2)) * 100, 2)
    })

df_comparison = pd.DataFrame(comparison_data)

print("\n" + "="*100)
print("TABLA COMPARATIVA DE CARACTERÍSTICAS ENTRE AMBAS IMÁGENES")
print("="*100)
print(df_comparison.to_string(index=False))
print("="*100)

# Visualizar la comparación
fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# Gráfico de barras para características seleccionadas
characteristics_to_plot = ['Area', 'Perimetro', 'Compacidad', 'Solidez', 'Excentricidad', 'Extension']
values_img1 = [features[char] for char in characteristics_to_plot]
values_img2 = [features_2[char] for char in characteristics_to_plot]

x = np.arange(len(characteristics_to_plot))
width = 0.35

axes[0].bar(x - width/2, values_img1, width, label='Imagen 1', alpha=0.8)
axes[0].bar(x + width/2, values_img2, width, label='Imagen 2', alpha=0.8)
axes[0].set_xlabel('Características')
axes[0].set_ylabel('Valor')
axes[0].set_title('Comparación de Características Principales')
axes[0].set_xticks(x)
axes[0].set_xticklabels(characteristics_to_plot, rotation=45, ha='right')
axes[0].legend()
axes[0].grid(axis='y', alpha=0.3)

# Gráfico de radar para comparación general
categories = list(features.keys())
N = len(categories)

# Normalizar valores para mejor visualización
values_img1_norm = [features[cat] / max(features[cat], features_2[cat]) if max(features[cat], features_2[cat]) > 0 else 0 for cat in categories]
values_img2_norm = [features_2[cat] / max(features[cat], features_2[cat]) if max(features[cat], features_2[cat]) > 0 else 0 for cat in categories]

angles = [n / float(N) * 2 * pi for n in range(N)]
values_img1_norm += values_img1_norm[:1]
values_img2_norm += values_img2_norm[:1]
angles += angles[:1]

ax = plt.subplot(2, 1, 2, projection='polar')
ax.plot(angles, values_img1_norm, 'o-', linewidth=2, label='Imagen 1', color='blue')
ax.fill(angles, values_img1_norm, alpha=0.15, color='blue')
ax.plot(angles, values_img2_norm, 'o-', linewidth=2, label='Imagen 2', color='red')
ax.fill(angles, values_img2_norm, alpha=0.15, color='red')
ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, size=8)
ax.set_ylim(0, 1)
ax.set_title('Comparación Normalizada de Características (Gráfico Polar)')
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
ax.grid(True)

plt.tight_layout()
plt.show()

print("\n✓ Análisis comparativo completado")
```
